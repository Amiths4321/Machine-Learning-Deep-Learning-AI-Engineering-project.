```python
class MyNeuralNetwork(nn.Module):

    def __init__(self):

        super().__init__()

        self.layer1 = nn.Linear(2, 8)
        self.relu1 = nn.ReLU()

        self.layer2 = nn.Linear(8, 4)
        self.relu2 = nn.ReLU()

        self.output = nn.Linear(4, 1)

    def forward(self, x):

        x = self.layer1(x)
        x = self.relu1(x)

        x = self.layer2(x)
        x = self.relu2(x)

        x = self.output(x)

        return x
```

```python
class SteelDefectCNN(nn.Module):

    def __init__(self):
        super().__init__()

        self.features = nn.Sequential(

            # Convolution block 1
            nn.Conv2d(3, 16, kernel_size=3, padding=1),
            nn.ReLU(),
            nn.MaxPool2d(2),

            # Convolution block 2
            nn.Conv2d(16, 32, kernel_size=3, padding=1),
            nn.ReLU(),
            nn.MaxPool2d(2)
        )

        self.classifier = nn.Sequential(

            nn.Flatten(),

            nn.Linear(32 * 16 * 16, 64),
            nn.ReLU(),

            nn.Linear(64, 6)
        )

    def forward(self, x):

        x = self.features(x)

        x = self.classifier(x)

        return x
```

```python
import torch
import torch.nn as nn
import torch.optim as optim

import numpy as np
from torch.utils.data import TensorDataset, DataLoader, random_split
```

```python
def create_pattern(label, size=64):
    image = np.zeros((size, size), dtype=np.float32)

    if label == 0:
        image[::8, :] = 1.0           # Horizontal lines

    elif label == 1:
        image[:, ::8] = 1.0           # Vertical lines

    elif label == 2:
        for i in range(size):
            image[i, i] = 1.0         # Diagonal line

    elif label == 3:
        image[::8, ::8] = 1.0         # Dotted grid

    elif label == 4:
        image[20:44, 20:44] = 1.0     # Central square

    elif label == 5:
        image[::4, ::4] = 1.0         # Dense grid

    return image
```

```python
images = []
labels = []

for label in range(6):
    for _ in range(100):

        image = create_pattern(label)

        # Add a little noise to make examples different
        noise = np.random.normal(0, 0.08, image.shape)
        image = np.clip(image + noise, 0, 1)

        # Convert grayscale into 3 channels
        image = np.stack([image, image, image])

        images.append(image)
        labels.append(label)

X = torch.tensor(np.array(images), dtype=torch.float32)
y = torch.tensor(labels, dtype=torch.long)

print("Images:", X.shape)
print("Labels:", y.shape)
```

```python
dataset = TensorDataset(X, y)

train_size = 480
test_size = 120

train_dataset, test_dataset = random_split(
    dataset,
    [train_size, test_size],
    generator=torch.Generator().manual_seed(42)
)

train_loader = DataLoader(
    train_dataset,
    batch_size=32,
    shuffle=True
)

test_loader = DataLoader(
    test_dataset,
    batch_size=32,
    shuffle=False
)

print("Training images:", len(train_dataset))
print("Testing images:", len(test_dataset))
```

```python
class SteelDefectCNN(nn.Module):

    def __init__(self):
        super().__init__()

        self.features = nn.Sequential(
            nn.Conv2d(3, 16, 3, padding=1),
            nn.ReLU(),
            nn.MaxPool2d(2),

            nn.Conv2d(16, 32, 3, padding=1),
            nn.ReLU(),
            nn.MaxPool2d(2)
        )

        self.classifier = nn.Sequential(
            nn.Flatten(),
            nn.Linear(32 * 16 * 16, 64),
            nn.ReLU(),
            nn.Linear(64, 6)
        )

    def forward(self, x):
        x = self.features(x)
        return self.classifier(x)


model = SteelDefectCNN()
```

```python
loss_fn = nn.CrossEntropyLoss()

optimizer = optim.Adam(
    model.parameters(),
    lr=0.01
)
```

```python
epochs = 15

for epoch in range(epochs):

    model.train()

    total_loss = 0

    for images, labels in train_loader:

        # Forward propagation
        outputs = model(images)

        # Calculate loss
        loss = loss_fn(outputs, labels)

        # Clear old gradients
        optimizer.zero_grad()

        # Backpropagation
        loss.backward()

        # Update weights
        optimizer.step()

        total_loss += loss.item()

    average_loss = total_loss / len(train_loader)

    print(
        f"Epoch {epoch+1}/{epochs}, "
        f"Loss: {average_loss:.4f}"
    )
```

```python
model.eval()

correct = 0
total = 0

with torch.no_grad():

    for images, labels in test_loader:

        outputs = model(images)

        predictions = torch.argmax(outputs, dim=1)

        correct += (predictions == labels).sum().item()
        total += labels.size(0)

accuracy = 100 * correct / total

print(f"Test Accuracy: {accuracy:.2f}%")
```

```python
class_names = [
    "Horizontal Lines",
    "Vertical Lines",
    "Diagonal",
    "Dotted Grid",
    "Central Square",
    "Dense Grid"
]

model.eval()

sample_image, actual_label = test_dataset[0]

with torch.no_grad():
    output = model(sample_image.unsqueeze(0))

    probabilities = torch.softmax(output, dim=1)

    predicted_label = probabilities.argmax(dim=1).item()

print("Actual:", class_names[actual_label])
print("Predicted:", class_names[predicted_label])
print(
    "Confidence:",
    f"{probabilities[0, predicted_label].item() * 100:.2f}%"
)
```
